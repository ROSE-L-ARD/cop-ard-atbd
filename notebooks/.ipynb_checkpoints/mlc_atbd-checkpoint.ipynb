## Overview

The Normalized Radar Backscatter (NRB) product was the first among the Analysis Ready
Data (ARD), with its introduction by CEOS in 2017 and with its definition further refined in 2021
with the release of version 5.5 of the document [20]. Today, various product specifications
for radar products have been merged into the initial version of the ”Combined CEOS-ARD for
Synthetic Aperture Radar” [22]. This product exclusively contains backscatter values and undergoes
Radiometric Terrain Correction (RTC), providing Gamma-Nought as defined in [112],
which mitigates the impact of differing observation geometries. It is the most readily comprehensible
product, enabling direct visualization for visual interpretation, as well as deeper analysis.
Indeed, analysing backscatter time series can yield information regarding the observed object’s
type for classification purposes (land use, land cover), as well as its temporal variations,
enabling applications such as change detection, security applications, and vegetation growth
monitoring.
NRB products have gained significant advancements in recent years, driven by both technological
innovation and increasing demand for high-quality, precise data. In the scientific literature,
numerous studies have explored the potential of NRB products for a wide range of applications.
Sentinel-1 satellites, equipped with SAR instruments, provide continuous and systematic observations,
facilitating the generation of NRB products at unprecedented spatial and temporal
resolutions. A pioneering example of NRB product, derived from Sentinel-1 GRD data, is provided
within the ”Digital Earth Africa” program [128]. Concurrently, NRB products from other
missions, such as ALOS-2, have also been developed [62].
Looking ahead, upcoming satellite missions acquiring data systematically at L-Band (e.g., NISAR,
ROSE-L) and freely distributing data to the scientific community will enhance observational capabilities
and address various challenges associated with single radar band measurements.

## Algorithm description



### Phase estimation

The multilooked phase is going to be generated by interfering each incoming CSLC acquisition $y_n$ with a reference $z_n$ that we call “running reference” and averaging over a small neighbourhood $\Omega$:
$$
   \phi_n = \angle \sum_{(r,a) \in \Omega} z_n^*(r,a) y_n(r,a).
$$
At the very beginning the reference $z_n$ is going to be initialized with the first image in the CSCL stack (i.e. $z_1 = y_1$). Later it is going to be updated as described below.

### Running reference update
$$
z_{n+1} = \alpha z_n + y_n e^{-j\phi_n}.
$$

```python
import zarr
import numpy as np
```

```python
base_dir = "/media/data/SAR/L2/Naples/ASC/S4/"
```

```python
# selecting window sizes (rg, az) and mini_size (number of acquisitions after which the long-term calibration kicks-in)
w_rg = 10
w_az = 20
mini_size = 5
```

```python
# opening the CSLC dataset
ds = zarr.open(base_dir + "slcs.zarr")
cstack = ds.coreg.cstack
N = len(cstack)
```

```python
# choosing the indices of one window (normally we are going to iterate over all window indices)
i_rg = 100
i_az = 500
```

```python
# building indices to cut a slice for the selected window
rg_range = np.arange((i_rg-1)*w_rg+1,(i_rg)*w_rg+1)
az_range = np.arange((i_az-1)*w_az+1,(i_az)*w_az+1)
wnd_range = np.ix_(az_range, rg_range)
```

```python
# initializing references (internal variables)
current_reference = cstack[0][wnd_range].ravel()
past_reference = np.zeros_like(current_reference)
```

```python
# inizializing output arrays
interf = np.zeros(N, dtype=np.complex64)
short_term_coherence = np.zeros(N)*np.nan
long_term_coherence = np.zeros(N)*np.nan
```

```python
def make_interf(A, B):
    return np.conj(A).dot(B)

def make_coh(A, B):
    return make_interf(A,B) / np.sqrt(make_interf(A,A) * make_interf(B,B))
```

```python
step = 0
for n in range(0,N):
    print(n)
    step = step + 1
    slc = cstack[n][wnd_range].ravel()  # getting the currect slc
    interf[n] = make_coh(current_reference,slc)
    short_term_coherence[n] = abs(interf[n])
    
    # update the current reference
    current_reference = current_reference*0.8 + slc*np.conj(interf[n])

    # generate (a) the past reference
    if step == mini_size :
                past_reference[:] = current_reference
        
    # long-term phase calibration of current_reference
    if step > mini_size :
                corr_phase = make_coh(past_reference,current_reference)
                if np.abs(corr_phase) > 0.1 :
                    current_reference[:] = current_reference * np.conj(corr_phase)/np.abs(corr_phase)
                    long_term_coherence[n] = np.abs(make_coh(past_reference,slc))
```
