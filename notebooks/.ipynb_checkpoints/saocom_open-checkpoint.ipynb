# Open Dataset

```python
# ds = datatree.open_datatree('/scratch/SAOCOM/S1A.zarr', engine='zarr', chunks={})
ds = datatree.open_datatree('/tudelft.net/staff-umbrella/Triplets4weu/SAOCOM/SAOCOM/CSLC_SAMPLE_PRODUCT/S1A.zarr', engine='zarr', chunks={})
```

## amplitude extraction

```python
ampl_ml8x4 = np.sqrt((np.abs(ds.orig['2024-08-08'].slc)**2).coarsen(azimuth_time=8, slant_range_time=4, boundary='trim').mean())
ampl_ml8x4.plot(vmax=1)
```

# Coregistered data

```python
ds.coreg.cstack
```

# Interferogram generation

```python
coreg = ds.coreg
```

```python
interf = (coreg.cstack.sel(acquisition_date='2024-08-08') * coreg.cstack.sel(acquisition_date='2024-07-15').conj()).coarsen(azimuth_time=8, slant_range_time=4, boundary='trim').mean()
phase = xr.apply_ufunc(np.angle, interf, dask='parallelized')
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
phase.plot()
```

```python
amp1 = (np.abs(coreg.cstack.sel(acquisition_date='2024-07-15'))**2).coarsen(azimuth_time=8, slant_range_time=4, boundary='trim').mean()
amp2 = (np.abs(coreg.cstack.sel(acquisition_date='2024-08-08'))**2).coarsen(azimuth_time=8, slant_range_time=4, boundary='trim').mean()
```

```python
cohe = np.abs(interf) / np.sqrt(amp1*amp2)
cohe.plot()
```

# COREGISTERED IMAGE SPECTRA

```python
import xrft
```

```python
fcoreg = xrft.fft(coreg.cstack.drop_vars(['azimuth_m', 'azimuth_time']).chunk({'azimuth_time': -1}), true_phase=True, true_amplitude=True, dim='azimuth_time')
spectr_az = np.abs(fcoreg).mean(dim='slant_range_time')
spectr_az[0].plot(label='slave')
spectr_az[1].plot(label='master')
plt.legend()
```

```python
fcoreg = xrft.fft(coreg.cstack.drop_vars('slant_range_m').chunk({'slant_range_time': -1}), true_phase=True, true_amplitude=True, dim='slant_range_time')
spectr_rg_crg = np.abs(fcoreg).mean(dim='azimuth_time')
spectr_rg_crg[0].plot(label='slave')
spectr_rg_crg[1].plot(label='master')
plt.legend()
```

```python
fcoreg = xrft.fft(coreg.cstack.drop_vars('slant_range_m').chunk({'slant_range_time': -1}), true_phase=True, true_amplitude=True, dim='slant_range_time')
spectr_rg_crg = np.abs(fcoreg).mean(dim='azimuth_time')
spectr_rg_crg[0].plot(label='coreg-slave')
fcoreg = xrft.fft(ds.orig["2024-07-15"].slc.drop_vars('slant_range_m').chunk({'slant_range_time': -1}), true_phase=True, true_amplitude=True, dim='slant_range_time')
spectr_rg_crg = np.abs(fcoreg).mean(dim='azimuth_time')
spectr_rg_crg[0].plot(label='orig-slave')
plt.legend()
```

```python
fcoreg = xrft.fft(coreg.cstack.drop_vars('slant_range_m').chunk({'slant_range_time': -1}), true_phase=True, true_amplitude=True, dim='slant_range_time')
spectr_rg_crg = np.abs(fcoreg).mean(dim='azimuth_time')
spectr_rg_crg[1].plot(label='coreg-master')
fcoreg = xrft.fft(ds.orig["2024-08-08"].slc.drop_vars('slant_range_m').chunk({'slant_range_time': -1}), true_phase=True, true_amplitude=True, dim='slant_range_time')
spectr_rg_crg = np.abs(fcoreg).mean(dim='azimuth_time')
spectr_rg_crg[0].plot(label='orig-master')
plt.legend()
```
